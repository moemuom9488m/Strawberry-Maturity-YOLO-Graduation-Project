```python
from bing_image_downloader import downloader
import os

# --- 設定 ---

# 1. 你要的搜尋關鍵字 (這是最重要的部分！)
# 我們需要各種真實情境，而不只是商品照
search_queries = [
    "strawberry ripening stages on plant", # 植株上的成熟階段
    "strawberry farm harvest",             # 草莓農場 採收
    "unripe green strawberry",             # 未成熟的綠草莓
    "white strawberry unripe",             # 未熟的白草莓
    "strawberry plant",                    # 草莓植株 (會包含花、果)
    "strawberries in market",              # 市場上的草莓
    "strawberries in punnet",              # 盒子裡的草莓
    "strawberry disease",                  # 這能幫你找到一些"壞"樣本 (可選)
    "overripe strawberry",                 # 過熟的草莓
]

# 2. 每個關鍵字要下載多少張圖片
# 建議先從 100-150 張開始，看情況再增加
limit_per_query = 100 

# 3. 圖片要儲存的主資料夾名稱
main_output_dir = "strawberry_dataset"

# --- 執行下載 ---

print(f"開始下載圖片，將儲存於 {main_output_dir} 資料夾中...")

for query in search_queries:
    print(f"\n--- 正在搜尋關鍵字: '{query}' (最多 {limit_per_query} 張) ---")
    
    downloader.download(
        query,
        limit=limit_per_query,
        output_dir=main_output_dir,
        adult_filter_off=True,
        force_replace=False,
        timeout=60,  # 增加超時時間，防止下載過慢
        verbose=True # 顯示詳細下載過程
    )

print("\n--- 所有下載任務完成！ ---")
print(f"請檢查 '{main_output_dir}' 資料夾。")
print("下一步：請手動清理這些圖片，移除不相關的內容。")
```

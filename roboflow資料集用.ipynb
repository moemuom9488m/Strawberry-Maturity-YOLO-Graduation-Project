```python
#合併所有圖片和標記檔到一個資料夾，並加入classes.txt
import os
import shutil

def merge_everything_to_one_folder(source_base, output_folder):
    # 設定支援的圖片格式
    img_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.webp')
    splits = ['train', 'val', 'test']
    
    # 建立單一目標目錄
    os.makedirs(output_folder, exist_ok=True)

    total_images = 0
    total_labels = 0

    print(f"開始從 {source_base} 提取檔案...")

    for split in splits:
        split_path = os.path.join(source_base, split)
        if not os.path.exists(split_path):
            continue
            
        img_src_dir = os.path.join(split_path, 'images')
        lbl_src_dir = os.path.join(split_path, 'labels')

        if os.path.exists(img_src_dir):
            for img_name in os.listdir(img_src_dir):
                if img_name.lower().endswith(img_extensions):
                    # 複製圖片
                    shutil.copy2(
                        os.path.join(img_src_dir, img_name),
                        os.path.join(output_folder, img_name)
                    )
                    total_images += 1
                    
                    # 尋找對應標記檔並複製到同一個資料夾
                    label_name = os.path.splitext(img_name)[0] + '.txt'
                    lbl_path = os.path.join(lbl_src_dir, label_name)
                    
                    if os.path.exists(lbl_path):
                        shutil.copy2(lbl_path, os.path.join(output_folder, label_name))
                        total_labels += 1

    # --- 關鍵修改：符合 LabelImg 格式的 classes.txt ---
    # 每一行代表一個 Index (0, 1, 2...)
    classes = ["level_1", "level_2", "level_3"]
    classes_file_path = os.path.join(output_folder, 'classes.txt')
    
    with open(classes_file_path, 'w', encoding='utf-8') as f:
        for cls in classes:
            f.write(cls + '\n')
    
    print(f"\n已建立 LabelImg 專用標籤檔: {classes_file_path}")
    # ----------------------------------------------

    print(f"\n--- 統計報告 ---")
    print(f"位置: {output_folder}")
    print(f"圖片總數: {total_images}")
    print(f"標記檔總數: {total_labels}")
    
    if total_images != total_labels:
        print(f"注意：圖片與標記檔數量不一致！(差額: {abs(total_images - total_labels)})")

# --- 修改此處 ---
source_dir = 'Maturity.v7i.yolov11' 
target_dir = 'roboflow_側拍_2'

merge_everything_to_one_folder(source_dir, target_dir)
```

```python
#處理 class 互換

import os

# 📁 你的 labels 資料夾（改成你的路徑）
LABELS_DIR = "roboflow_側拍_2\\綜合"

# 📸 處理的圖片範圍
START = 1
END = 1021

def swap_classes_in_label(file_path):
    new_lines = []

    with open(file_path, "r") as f:
        lines = f.readlines()

    for line in lines:
        parts = line.strip().split()
        if len(parts) < 5:
            continue

        cls = int(parts[0])

        # 🔄 類別互換
        if cls == 0:
            cls = 2
        elif cls == 2:
            cls = 0

        parts[0] = str(cls)
        new_lines.append(" ".join(parts))

    # ✍️ 寫回檔案
    with open(file_path, "w") as f:
        f.write("\n".join(new_lines) + "\n")


# 🚀 主程式
for i in range(START, END + 1):
    filename = f"{i}.txt"
    filepath = os.path.join(LABELS_DIR, filename)

    if os.path.exists(filepath):
        swap_classes_in_label(filepath)
        print(f"✅ 已處理: {filename}")
    else:
        print(f"⚠️ 找不到: {filename}")

print("\n🎉 完成 class 互換！")
```

```python
# 分析 Roboflow 資料集的標註衝突
import os
import cv2
from ultralytics import YOLO
from pathlib import Path

# --- 設定路徑 ---
MODEL_PATH = "runs/detect/train/weights/best.pt"  # 你剛練好的最強模型
ROBOFLOW_IMG_DIR = "roboflow_merged_split/train/images"  # Roboflow 圖片路徑
ROBOFLOW_LBL_DIR = "roboflow_merged_split/train/labels"  # Roboflow 標籤路徑
OUTPUT_LOG = "annotation_conflicts.txt"

# 載入模型
model = YOLO(MODEL_PATH)
class_names = ['level_1', 'level_2', 'level_3']

conflicts = []

print(f"🚀 開始分析 Roboflow 資料集衝突...")

# 遍歷 Roboflow 圖片
img_paths = list(Path(ROBOFLOW_IMG_DIR).glob("*.jpg"))

for i, img_path in enumerate(img_paths):
    name = img_path.stem
    lbl_path = os.path.join(ROBOFLOW_LBL_DIR, f"{name}.txt")
    
    if not os.path.exists(lbl_path):
        continue

    # 1. 讀取 Roboflow 的原始標註 (Ground Truth)
    gt_classes = []
    with open(lbl_path, 'r') as f:
        for line in f:
            gt_classes.append(int(line.split()[0]))

    # 2. 執行模型預測 (Inference)
    results = model.predict(source=str(img_path), conf=0.5, save=False, verbose=False)
    pred_classes = results[0].boxes.cls.cpu().numpy().astype(int).tolist()

    # 3. 衝突邏輯判斷 (簡單邏輯：如果預測出的類別組合跟標註完全對不起來)
    # 這裡採用的檢查方式是：檢查預測出的主導類別是否在標註中出現
    if pred_classes:
        # 如果模型預測的類別，在原始標註裡完全沒出現過，視為衝突
        is_conflict = False
        for p in set(pred_classes):
            if p not in set(gt_classes):
                is_conflict = True
                break
        
        # 嚴重的「跨級」衝突檢查 (例如 1 誤判為 3)
        if 0 in gt_classes and 2 in pred_classes:
            is_conflict = True # 標註說有青果，模型卻只看到熟果
            
        if is_conflict:
            conflicts.append(f"Conflict in {name}: GT={gt_classes}, PRED={pred_classes}")

    if (i + 1) % 100 == 0:
        print(f"進度: {i+1}/{len(img_paths)}")

# 儲存結果
with open(OUTPUT_LOG, 'w') as f:
    f.write("\n".join(conflicts))

print(f"\n✅ 分析完成！共發現 {len(conflicts)} 張潛在標註衝突圖片。")
print(f"請查看 {OUTPUT_LOG} 以進行手動抽查。")
```

```python
import os
import shutil
import random
from pathlib import Path

# --- 設定路徑 ---
SRC_DIR = "./roboflow_merged"       # 原始 Roboflow 資料夾
DST_DIR = "./roboflow_1000_sample"  # 抽取後的存放路徑
SAMPLE_SIZE = 1000                  # 抽取數量

def extract_random_sample_with_classes(src, dst, size):
    # 1. 建立目標資料夾（如果存在就先刪除，確保資料純淨）
    if os.path.exists(dst):
        shutil.rmtree(dst)
    os.makedirs(dst, exist_ok=True)

    # 2. 優先處理 classes.txt
    classes_file = Path(src) / "classes.txt"
    if classes_file.exists():
        shutil.copy2(classes_file, Path(dst) / "classes.txt")
        print(f"📄 已成功複製 classes.txt")
    else:
        print(f"⚠️ 警告：在 {src} 中找不到 classes.txt！")

    # 3. 獲取所有圖片列表
    image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
    all_images = [f for f in Path(src).iterdir() if f.suffix.lower() in image_extensions]
    
    if len(all_images) < size:
        print(f"❌ 錯誤：圖片總數 ({len(all_images)}) 不足 {size} 張")
        return

    # 4. 隨機抽取
    sample_images = random.sample(all_images, size)
    
    print(f"🚀 開始從 {len(all_images)} 張中抽取 {size} 張圖片與標籤...")
    
    success_count = 0
    for img_path in sample_images:
        name = img_path.stem
        txt_name = f"{name}.txt"
        txt_path = Path(src) / txt_name
        
        # 複製圖片
        shutil.copy2(img_path, Path(dst) / img_path.name)
        
        # 複製對應標籤
        if txt_path.exists():
            shutil.copy2(txt_path, Path(dst) / txt_name)
            success_count += 1
        else:
            print(f"❓ 找不到圖片 {img_path.name} 對應的標籤檔")
            
    print(f"\n✅ 抽取完成！")
    print(f"📍 存放路徑: {dst}")
    print(f"📊 成功抽取組數: {success_count} 組 (圖+標籤)")

# 執行抽取
if __name__ == "__main__":
    extract_random_sample_with_classes(SRC_DIR, DST_DIR, SAMPLE_SIZE)
```
